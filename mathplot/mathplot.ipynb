# Mathplotlib Line flotes

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,0.1)
y=2*x+5

plt.plot(x,y)
plt.show()


```

# Custemiizing The Line Plot

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,1)
y=2*x+5

plt.plot(x,y,linewidth=2.5,color='red',linestyle=':',alpha=0.7,marker='o',markersize=8,markerfacecolor='blue',markeredgecolor='green')
plt.title("Line Plot Demo")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend(["Line Plot"],loc='best')
plt.show()
```

# two diffrent lines

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,1)
y1=2*x+5
y2=3*x+10

plt.subplot(1,2,1)
plt.plot(x,y1,linewidth=2.5,color='red',linestyle=':',alpha=0.7,marker='o',markersize=8,markerfacecolor='blue',markeredgecolor='green')
plt.title("Graph 1")
plt.subplot(1,2,2)
plt.plot(x,y2,linewidth=2.5,color='orange',linestyle='--',alpha=0.7,marker='s',markersize=8,markerfacecolor='yellow',markeredgecolor='purple')
plt.title("Graph 2")
plt.show()
```

# Bar plot

```python
import numpy as np
import matplotlib.pyplot as plt

data={'apples':20,'oranges':15,'grapes':30,'bananas':25}
names=list(data.keys())
values=list(data.values())
fig=plt.figure(figsize=(10,5))
plt.bar(names,values,color=['red','orange','purple','yellow'])
plt.title("Bar Chart Demo")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.show()
```
