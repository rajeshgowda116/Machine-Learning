# DateTime Module

To Get Date

```python
import datetime as dt
d=dt.date(2026,7,11)
print(d)
```

To Get Todays Date

```python
today=dt.date.today()
print(today)
print(today.day)
print(today.month)
print(today.year)
```

To get Week Day

```python
#today is monday thats why that shows like that
print(today.weekday())
print(today.isoweekday())
```

Add substract dates

```python
tdelta=dt.timedelta(days=7)
print(today-tdelta)
print(today+tdelta)
```

Find brith date

```python
today=dt.date.today()
bday=dt.date(2005,12,4)
age=today-bday
print(age.total_seconds())
print(age.days)
```

# Time

```python
tim=dt.time(9,30,45)
print(tim)
print(tim.hour)
print(tim.minute)
print(tim.second)
```
