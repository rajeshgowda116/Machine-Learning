```python
import os
if(not os.path.exists("data")):
   os.mkdir("data")
for i in range(0,100):
  if(not os.path.exists(f"data/day{i}")):
    os.mkdir(f"data/day{i}")
```

```python
for i in range(100):
  os.rename(f"data/tuto{i}",f"data/tuto {i}")
```

```python
folders=os.listdir("data")
for folder in folders:
  print(folder)
```

```python
files=os.listdir("data")
for file in files:
  print(file)
  print(os.listdir(f"data/{file}"))
```
