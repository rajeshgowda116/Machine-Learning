# Mathplotlib Line flotes

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,0.1)
y=2*x+5

plt.plot(x,y)
plt.show()


```

# Custemiizing The Line Plot

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,1)
y=2*x+5

plt.plot(x,y,linewidth=2.5,color='red',linestyle=':',alpha=0.7,marker='o',markersize=8,markerfacecolor='blue',markeredgecolor='green')
plt.title("Line Plot Demo")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend(["Line Plot"],loc='best')
plt.show()
```

# two diffrent lines

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,10,1)
y1=2*x+5
y2=3*x+10

plt.subplot(1,2,1)
plt.plot(x,y1,linewidth=2.5,color='red',linestyle=':',alpha=0.7,marker='o',markersize=8,markerfacecolor='blue',markeredgecolor='green')
plt.title("Graph 1")
plt.subplot(1,2,2)
plt.plot(x,y2,linewidth=2.5,color='orange',linestyle='--',alpha=0.7,marker='s',markersize=8,markerfacecolor='yellow',markeredgecolor='purple')
plt.title("Graph 2")
plt.show()
```

# Bar plot

```python
import numpy as np
import matplotlib.pyplot as plt

data={'apples':20,'oranges':15,'grapes':30,'bananas':25}
names=list(data.keys())
values=list(data.values())
fig=plt.figure(figsize=(10,5))
plt.bar(names,values,color=['red','orange','purple','yellow'])
plt.title("Bar Chart Demo")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data={'apples':20,'oranges':15,'grapes':30,'bananas':25}
names=list(data.keys())
values=list(data.values())
fig=plt.figure(figsize=(10,5))
plt.barh(names,values,color=['red','orange','purple','yellow'])
plt.title("Bar Chart Demo")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.show()
```

# Scater plot

```python
import numpy as np
import matplotlib.pyplot as plt
a=[10,20,30,40,50,60,70,80]
b=[1,2,5,9,7,5,3,2]
x=[1,2,3,4,5,6,7,8]
plt.scatter(a,b,color='red',marker='o',s=100,label='Dataset A')
plt.scatter(a,x,color='blue',marker='s',s=100,label='Dataset B')
plt.legend(loc='best')
plt.title("Scatter Plot Demo")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")  
plt.grid(True)
plt.savefig("scatter_plot.png")
plt.show()
```

# Histogram

```python
import matplotlib.pyplot as plt

numbers=[10,12,15,18,20,22,26,29,30,32,36,39,45,60,68,63,70,78,90,95,99]
plt.hist(numbers,bins=[0,20,40,60,80,100],color='green',edgecolor='black')
plt.title("Histogram Demo")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")  
plt.legend(["Numbers"],loc='upper right')
plt.savefig("histogram.png")
plt.show()
```
