# Aggregate Function

```python
a=np.array([1,2,3,4,5,6])
print("sum:",np.sum(a))
print("mean:",np.mean(a))
print("max:",np.max(a))
print("min:",np.min(a))
print("standard deviation:",np.std(a))
print("variance:",np.var(a))
print("median:",np.median(a))
print("coorelation coefficient:",np.corrcoef(a))
```
