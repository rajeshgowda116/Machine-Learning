# merze operation

```python
# merze operation
player=['player1','player2','player3']
points=[8,9,6]
title=['game1','game2','game3']
df1=pd.DataFrame({'player':player,'points':points})
print(df1)
df2=pd.DataFrame({'player':player,'title':title})
print(df2)
df3=pd.merge(df1,df2,on='player')
df3
```

```python
# merze operation
player=['player1','player2','player3']
points=[8,9,6]
title=['game1','game2','game3']
df1=pd.DataFrame({'player':player,'points':points,'title':title})
df1
```

```python
player=['player1','player5','player6']
power=['punch','kick','block']
title=['game1','game5','game6']
df2=pd.DataFrame({'player':player,'power':power,'title':title})
df2
```

# innermerze

```python
df3=pd.merge(df1,df2,on='player',how='inner')
df3
```

```python
df4=pd.merge(df1,df2,on='title',how='inner')
df4
```

Right merze

```python
df6=pd.merge(df1,df2,on='player',how='right')
df6
```

# Join oparation

```python
player=['player1','player5','player6']
power=['punch','kick','block']
title=['game1','game5','game6']
df10=pd.DataFrame({'player':player,'power':power,'title':title},index=['l1','l2','l3'])
df10
```

```python
player=['player1','player5','player6']
power=['punch','kick','block']
title=['game1','game5','game6']
df12=pd.DataFrame({'player':player,'power':power,'title':title},index=['l2','l3','l4'])
df12
```

inner join

```python
# inner join with suffixes to avoid overlapping column names
# original error occurred because both DataFrames shared identical column names

result = df10.join(df12, how='inner', lsuffix='_10', rsuffix='_12')
result
```

```python
df3=df10.join(df12, how='left', lsuffix='_10', rsuffix='_12')
df3
```

```python
df3=df10.join(df12, how='right', lsuffix='_10', rsuffix='_12')
df3
```
