# 1️⃣ Attrition Analysis

```python
import pandas as pd
import openpyxl
HR=pd.read_excel("HRData.xlsx")
HR
```

```python
HR.info
```

```python
HR.isnull().sum()
```

```python
HR.describe()
```

```python
Att=HR['Attrition'].value_counts()
Att
```

```python
import matplotlib.pyplot as plt
plt.pie(Att,labels=['NO','Yes'],autopct='%1.1f%%')
plt.title(" Attrition Analysis")
plt.show()
```

```python
dept=dict(HR.groupby('Attrition')['Department'].value_counts())
dept
```

```python
lables=[f'{i[0]}-{i[1]}' for i in dept.keys()]
values=list(dept.values())
plt.figure(figsize=(8,5))
plt.bar(lables,values,color=['green','green','green','red','red','red'])
plt.xlabel("Attrition - Department")
plt.ylabel("Number of Employees")
plt.title("Employee Attrtion by Department")
plt.show()
```

```python
gender=dict(HR.groupby('Attrition')['Gender'].value_counts())
gender
```

```python
labels1=[f"{i[0]}-{i[1]}" for i in gender.keys()]
values1=list(gender.values())
plt.bar(labels1,values1,color=['green','green','red','red'])
plt.xlabel("Attrition - Department")
plt.ylabel("Number of Employees")
plt.title("Employee Attrtion by Department")
plt.show()
```

```python
plt.hist(HR['Monthly Income'],bins=10,edgecolor='black',color='red',alpha=0.7)
plt.xlabel("Monthly Salary")
plt.ylabel("Number of Employees")
plt.title("Salary Distribution of Employees")
plt.tight_layout()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
year=HR.groupby('Attrition')['Total Working Years'].value_counts()
year=dict(year['No'])
key=list(year.keys())
values=list(year.values())
plt.bar(key,values)
plt.title("Total Working Years of Employees (Attrition = No)")
plt.xlabel("Total Working Years")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.hist(HR['Age'],edgecolor='black')
plt.xlabel("Age")
plt.ylabel("Employees")
plt.title("Age Ratio of The Employees")
plt.show()
```
