# 2️⃣ IPL Data Analysis 🏏

```python
import pandas as pd
import seaborn as sns
ipl=pd.read_csv('ipl 2017.csv')
ipl.head()
```

```python
print(ipl.columns)
```

```python
ipl.info()
```

```python
ipl.isnull().sum()
```

```python
ipl.describe()
```

```python
winner=dict(ipl['Match Winner'].value_counts())
```

```python
import matplotlib.pyplot as plt
bars = plt.bar(winner.keys(),winner.values(),color=['blue','orange','green','red','purple','brown','pink','red'])
plt.xticks(rotation=90)
plt.bar_label(bars) 
plt.xlabel("Teams")
plt.ylabel("Number of Matches Won")
plt.title("Number of Matches Won by Each Team")
plt.show()
```

```python
man_of_the_match=ipl["Player of the match"].value_counts()
print(man_of_the_match)
```

```python
top=man_of_the_match.head(5)
print("Top 5 Players of the Match:", top)
```

```python
team_mom = ipl["Match Winner"].value_counts()
print(team_mom)
```

```python
team_mom = ipl.groupby("Match Winner")["Player of the match"].count()
print(team_mom)
team_mom1 = dict(team_mom)
```

```python
plt.bar(team_mom1.keys(),team_mom1.values(),color=['blue','orange','green','red','purple'])
plt.xticks(rotation=90)
plt.xlabel("Teams")
plt.ylabel(" Number of man of the match winners")
plt.title("Team With Most Man of the Match Winners")
plt.show()
```

```python
win=ipl["Toss Winner"]==ipl["Match Winner"]
print(win.value_counts())
```

```python
perce=win.value_counts(normalize=True)*100
print(perce)
```

```python
plt.pie(perce, labels=['Won','Lose'], autopct='%1.1f%%', colors=['lightblue', 'lightcoral'])
plt.title("Percentage of Matches Won by Toss Winner vs Toss Loser")
plt.show()
```
