```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
students =pd.read_csv('student_marks.csv')
students
```

```python
students['Total']=students['Maths']+students['Physics']+students['Chemistry']+students['English']+students['Biology']+students['Economics']+students['History']+students['Civics']
students['Average']=students['Total']/8
students
```

```python
highest_marks=students[['Maths','Physics','Chemistry','English','Biology','Economics','History','Civics']].max()
highest_marks
```

```python
lowest_marks=students[['Maths','Physics','Chemistry','English','Biology','Economics','History','Civics']].min()
lowest_marks
```

```python
highest_average=students['Average'].max()
highest_average
```

```python
lowest_avaerage=students['Average'].min()
lowest_avaerage
```

Grade System

```python
conditions = [
    students["Average"] >= 90,
    students["Average"] >= 75,
    students["Average"] >= 60,
    students["Average"] < 60
]
choices = ['A', 'B', 'C', 'D']
# provide a string default so dtype matches the choices
students['Grade'] = np.select(conditions, choices, default='D')
students
```

# Graph

```python
plt.figure(figsize=(10,6))
plt.bar(students['name'], students['Average'], color='skyblue')
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
grades_count = students['Grade'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(grades_count.index, grades_count.values, color='salmon')
plt.xlabel('Grades')
plt.ylabel('Number of Students') 
plt.title('Distribution of Grades')
plt.show()
```
